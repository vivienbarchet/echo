```python
import pandas as pd 
import numpy as np 
import pickle
```

```python

unique_df = pd.read_csv("sentences_ids.csv")
```

```python
import string



# Map lowercase letters 'a' through 'z' to indices 0 to 25
ALPHABET = string.ascii_lowercase
CHAR_TO_IDX = {char: i for i, char in enumerate(ALPHABET)}


def get_first_letter_matrix(text):
    words = text.split()
    # Create a zero matrix of shape (num_words, 26)
    matrix = np.zeros((len(words), len(ALPHABET)), dtype=int)

    for i, word in enumerate(words):
        first_char = word[0].lower()
        if first_char in CHAR_TO_IDX:
            matrix[i, CHAR_TO_IDX[first_char]] = 1

    return matrix


# Store the 2D NumPy array in the third column
unique_df["first_letter_embeddings"] = unique_df["text"].apply(get_first_letter_matrix)

print(unique_df["first_letter_embeddings"].iloc[0])
# Output shape: (2, 26) -> Row 0 has 1 at index 7 ('h'), Row 1 has 1 at index 22 ('w')
```

```python
with open("./embeddings/embeddings_onehot.pickle", "wb") as o:
    pickle.dump(unique_df, o)
```
